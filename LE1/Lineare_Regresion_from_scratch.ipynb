```python
import numpy as np
import pandas as pd

# Zufallsseed für Reproduzierbarkeit
np.random.seed(42)

# 100 zufällige x-Werte zwischen -50 und 50
x = np.random.uniform(-50, 50, 100).reshape(-1, 1)

# lineare Funktion mit Rauschen
y = 3.2 * x - 10 + np.random.normal(5, 30, size=(100, 1))

# In DataFrame packen (optional)
df = pd.DataFrame({"x": x.flatten(), "y": y.flatten()})
```

```python
def mean_square_error(m, b, points):
    total_error = 0
    for i in range(len(points)):
        x = points.iloc[i].AbsetnHours
        y = points.iloc[i].Age
        total_error +=(y-(m*x+b))**2
    total_error/ float(len(points))
```

# Mit dieser Funktion können Lineare Regressionsgerade berechnet werden

```python
def gradient_descent(m_now, b_now, points, L):
    m_gradient = 0
    b_gradient = 0

    n = len(points)

    for i in range(n):
        x = points.iloc[i].x #x an name der Daten anpassen
        y = points.iloc[i].y #y an Name der Daten anpassen

        m_gradient += -(2/n) * x * (y - (m_now * x + b_now))
        b_gradient += -(2/n) * (y - (m_now * x + b_now))

    m = m_now - m_gradient * L
    b = b_now - b_gradient * L
    return m, b
```

```python
m = 0
b = 0
L = 0.0001
epochs = 300

for i in range(epochs):
    m, b = gradient_descent(m, b, df, L, x, y)

print(m, b)

plt.scatter(y=df.y, x=df.x)
plt.plot(list(range(-50, 50)), [m * x + b for x in range(-50,50)], color='red')
```
